```python
from ultralytics import YOLO
```

```python
model = YOLO()
```

```python
import cv2
```

```python
cam = cv2.VideoCapture('car_video_3.mp4')
while True:
    ret, video = cam.read()
    if not ret:
        print('video not read')
        break
    pre = model.predict(video)
    annoted_video = pre[0].plot()
    cv2.imshow('license_plate_detector', annoted_video)
    if cv2.waitKey(1) == ord('q'):
        break

cam.release()
cv2.destroyAllWindows()
```

```python
from PIL import Image
def image(image):
    predicted = model.predict(image)
    annoted_image = predicted[0].plot()
    return Image.fromarray(annoted_image)
```

```python
import gradio as gr
```

```python
app = gr.Interface(fn=image,
    inputs=gr.Image(type="numpy"),
    outputs=gr.Image(type="pil"),
    title="License Plate Detection",
    description="Upload an image to detect license plates using YOLO.")
```

```python
app.launch(show_error=True)
```
